#Data Cleaning And Preprocessing

```python
df_cleaned = df.copy()
df_cleaned.head()
```

```python
df_cleaned.shape
```

```python
df_cleaned.drop_duplicates(inplace=True)
df_cleaned.shape
```

```python
df_cleaned.dtypes
```

```python
df_cleaned['sex'].value_counts()
```

```python
df_cleaned['sex'] = df_cleaned['sex'].map({'male':0, 'female':1})
df_cleaned.head()
```

```python
df_cleaned['smoker'].value_counts()
```

```python
df_cleaned['smoker'] = df_cleaned['smoker'].map({'yes': 1, 'no': 0})
df_cleaned.head()
```

```python
df_cleaned.rename(columns={
    'sex': 'is_female',
    'smoker': 'is_smoker'
}, inplace=True)
df_cleaned.head()
```

```python
df_cleaned['region'].value_counts()
```

```python
df_cleaned = pd.get_dummies(df_cleaned, columns=['region'], drop_first=False)
df_cleaned.head()
```

```python
df_cleaned = df_cleaned.astype(int)
df_cleaned.head()
```

#Feature Engineering and Extraction

```python
df_cleaned['bmi_category'] = pd.cut(
                    df_cleaned['bmi'], 
                    bins=[0, 18.5, 24.9, 29.9, np.inf], 
                    labels=['Underweight', 'Normal weight', 'Overweight', 'Obese']
                )
df_cleaned.head()
```

```python
df_cleaned = pd.get_dummies(df_cleaned, columns=['bmi_category'], drop_first=False)
df_cleaned.head()
```

```python
df_cleaned = df_cleaned.astype(int)
```

```python
df_cleaned.head()
```

```python
from sklearn.preprocessing import StandardScaler
cols = ['age', 'bmi', 'children']
scaler = StandardScaler()

df_cleaned[cols] = scaler.fit_transform(df_cleaned[cols])
df_cleaned.head()
```

```python
from scipy.stats import pearsonr

# pearson correlation 

selected_features = ['age', 'bmi', 'children', 'is_female', 'is_smoker']

correlations = {
    feature: pearsonr(df_cleaned[feature], df_cleaned['charges'])[0] 
    for feature in selected_features
}

correlation_df = pd.DataFrame(list(correlations.items()), columns=['Feature', 'Pearson Correlation'])
correlation_df.sort_values(by='Pearson Correlation', ascending=False)
```

```python
cat_features = ['is_female', 'is_smoker', 'region_northwest', 'region_southeast', 'region_southwest', 'bmi_category_Underweight', 'bmi_category_Normal weight', 'bmi_category_Overweight', 'bmi_category_Obese']  
```

```python
from scipy.stats import chi2_contingency
import pandas as pd

alpha = 0.05

df_cleaned['charges_bin'] = pd.qcut(df_cleaned['charges'], q=4, labels=False)
chi2_results = {}

for col in cat_features:
    contingency = pd.crosstab(df_cleaned[col], df_cleaned['charges_bin'])
    chi2_stat, p_val, _, _ = chi2_contingency(contingency)
    
    decision = 'Reject Null (Keep Feature)' if p_val < alpha else 'Accept Null (Drop Feature)'
    
    chi2_results[col] = {
        'chi2_statistic': chi2_stat,
        'p_value': p_val,
        'Decision': decision
    }

chi2_df = pd.DataFrame(chi2_results).T
chi2_df.sort_values(by='p_value')
chi2_df
```

```python
final_df = df_cleaned[['age', 'bmi', 'children', 'is_smoker', 'charges','region_southeast', 'bmi_category_Obese']]
```

```python
final_df
```

```python
from sklearn.model_selection import train_test_split
x = final_df.drop('charges', axis=1)
y = final_df['charges']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(y_test, y_pred)
r2
```

```python
n = x_test.shape[0]
p = x_test.shape[1]
adjusted_r2 = 1 - (1 - r2) * (n - 1) / (n - p - 1)
adjusted_r2
```
